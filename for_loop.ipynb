```python
#Q3/10 — Find the first repeated character
word = "abcdefgab"
visited=[]
for i in word:
  if i not in visited:
    count=0
    for j in word:
      if i == j:
        count+=1
    visited.append(i)
    if count > 1:
      print(i)
      break
```
